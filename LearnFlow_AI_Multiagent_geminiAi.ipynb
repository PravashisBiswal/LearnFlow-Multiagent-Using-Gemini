```python
import os
from crewai import Agent, Task ,Crew,LLM
from crewai_tools import SerperDevTool


os.environ['GEMINI_API_KEY']="AQ.Ab8RN6KUdfcbD6EcRzY5-pqR1vePPNbKY1dN"
os.environ['SERPER_API_KEY']="79d400f5d631c49a61176725e870318ec"
search_tool=SerperDevTool()

# Instantiate the LLM explicitly
llm = LLM(
         model="gemini/gemini-3-flash-preview",
         api_key=os.environ['GEMINI_API_KEY'],
         temperature=0.5
         )

teacher=Agent(
    role="Teacher",
    goal="Explain concept clearly step by step ",
    backstory="experienced teacher who explains with examples",
    llm=llm,
    verbose=True
)

researcher=Agent(
    role="Researcher",
    goal="Find latest information",
    backstory="Expert in searching real-world data",
    llm=llm,
    tools=[search_tool],
    verbose=True
)



simplifier=Agent(
    role="Simplifier",
    goal="Make things easy",
    backstory="Breaks complex ideas ino simple language",
    llm=llm,
    verbose=True

)


student=Agent(
    role="Student",
    goal="Take notes",
    backstory="Writes simple notes like a learner",
    llm=llm,
    verbose=True

)

examiner=Agent(
    role="Examiner",
    goal="Create questions",
    backstory="Tests understanding",
    llm=llm,
    verbose=True
)

topic ="What is Artificial Intelligence "

task1=Task(
    description=f"Search and find three import points about{topic}",
    expected_output="3 clear points",
    agent=researcher

)

task2=Task(
    description=f"Explain{topic} in simple steps with examples",
    expected_output="step by step explaination",
    agent=teacher
)

task3=Task(
    description=f"Simplify the explaination of {topic}",
    expected_output="very easy  explaination",
    agent=simplifier
)

task4=Task(
    description=f"Write short notes on  {topic} like student",
    expected_output="Simple notes",
    agent=teacher
)


task5=Task(
    description=f"Create 3 simple questionson  {topic} ",
    expected_output="3 easy questions",
    agent=examiner
)

crew=Crew(
    agents=[teacher,researcher,simplifier,student,examiner],
    tasks=[task1,task2,task3,task4,task5],
    verbose=True
)

result = await crew.kickoff_async()
print(result)
```
